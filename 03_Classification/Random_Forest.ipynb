```python
from sklearn.datasets import make_classification
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score , classification_report
from sklearn.tree import DecisionTreeClassifier
import pandas as pd
import matplotlib.pyplot as plt
```

```python
X,y= make_classification(
    n_samples=10000,
    n_features=10,
    n_redundant=3,
    n_informative=5,
    random_state=40
)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y, test_size=0.2 , random_state=40)
```

```python
rf = RandomForestClassifier()
```

```python
rf.fit(X_train,y_train)
```

```python
y_pred = rf.predict(X_test)
```

```python
dt= DecisionTreeClassifier()
```

```python
dt.fit(X_train,y_train)
```

```python
y_pred2 = dt.predict(X_test)
```

```python
print("ACCURACY SCORE(RANDOM FOREST):",accuracy_score(y_test,y_pred))
print("CLASSIFICATION REPORT(RANDOM FOREST):",classification_report(y_test,y_pred))

print("ACCURACY SCORE (DECISION TREE):",accuracy_score(y_test,y_pred2))
print("CLASSIFICATION REPORT(DECISION TREE):",classification_report(y_test,y_pred2))
```

```python
for i in [10, 20, 50, 100, 200, 500]:
    rf = RandomForestClassifier(n_estimators=i, random_state=42)
    rf.fit(X_train, y_train)
    train_acc = accuracy_score(y_train, rf.predict(X_train))
    test_acc = accuracy_score(y_test, rf.predict(X_test))
    print(f"n_estimators={i} | Train: {train_acc:.4f} | Test: {test_acc:.4f}")
    
```

```python
importances = rf.feature_importances_
plt.bar(range(10), importances)
plt.title('Feature Importance')
plt.show()
```

## Key Findings

Random Forest vs Decision Tree:
- Decision Tree: 95% accuracy
- Random Forest: 98% accuracy
- 3% improvement from ensemble averaging

n_estimators sweet spot = 100
Adding more trees beyond 100 gives no accuracy gain but increases training time.

Feature 1 is the most important predictor in this dataset.

Why Random Forest beats Decision Tree:
- Multiple trees trained on random data subsets (bagging)
- Each tree sees random feature subsets (feature randomness)
- Averaging reduces variance without increasing bias

```python
import numpy as np

importances = rf.feature_importances_
indices = np.argsort(importances)[::-1]

plt.figure(figsize=(10, 5))
plt.bar(range(10), importances[indices])
plt.xticks(range(10), [f'Feature {i}' for i in indices], rotation=45)
plt.title('Feature Importance — Random Forest (sorted)')
plt.tight_layout()
plt.show()
```
