```python
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score , classification_report
from sklearn.preprocessing import StandardScaler
```

```python
X,y=make_classification(
    n_samples = 100000,
    n_features = 15,
    n_informative = 5,
    n_redundant= 3,
    random_state = 40
)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y, test_size=0.2 , random_state=40)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
lor = LogisticRegression()
lor.fit(X_train_scaled , y_train)
y_pred_lor = lor.predict(X_test)
```

```python
dt = DecisionTreeClassifier()
dt.fit(X_train,y_train)
y_pred_dt = dt.predict(X_test)
```

```python
rf = RandomForestClassifier()
rf.fit(X_train,y_train)
y_pred_rf = rf.predict(X_test)
```

```python
xgb = XGBClassifier()
xgb.fit(X_train,y_train)
y_pred_xgb = xgb.predict(X_test)
```

```python
print("ACCURACY SCORE" , accuracy_score(y_test, y_pred_lor))
print("ACCURACY SCORE" , accuracy_score(y_test, y_pred_dt))
print("ACCURACY SCORE" , accuracy_score(y_test, y_pred_rf))
print("ACCURACY SCORE" , accuracy_score(y_test, y_pred_xgb))
```

```python
import time
```

```python
models = {
    "Logistic Regression": LogisticRegression(random_state=40),
    "Decision Tree": DecisionTreeClassifier(random_state=40),
    "Random Forest": RandomForestClassifier(random_state=40),
    "XGBoost": XGBClassifier(random_state=40)
}
```

```python
for name, model in models.items():
    # Measure Training Time
    start_train = time.time()
    model.fit(X_train, y_train)
    train_time = time.time() - start_train
    
    # Measure Prediction Time
    start_pred = time.time()
    y_pred = model.predict(X_test)
    pred_time = time.time() - start_pred
    
    # Evaluate Accuracy
    acc = accuracy_score(y_test, y_pred)
    
    print(f"{name:20s} | Fit: {train_time:.4f}s | Predict: {pred_time:.4f}s | Test Acc: {acc:.4f}")
```

## Key Finding — Model Comparison

| Model | Accuracy | Training Time |
|-------|----------|--------------|
| Logistic Regression | 87% | 0.18s (fastest) |
| Decision Tree | 94.6% | 3.4s |
| Random Forest | 96.6% | 34.7s (slowest) |
| XGBoost | 96.5% | 5.4s |

Winner: XGBoost
- Nearly identical accuracy to Random Forest
- 6x faster training time
- Industry standard for tabular data for this reason

When to use each:
- Logistic Regression: quick baseline, interpretability needed
- Decision Tree: when explainability is critical
- Random Forest: when accuracy matters more than speed
- XGBoost: best accuracy + speed tradeoff for production
